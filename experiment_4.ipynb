# Experiment 4: Python Programming for Visualization

**Objective:** Visualize dataset distributions, relationships, and trends using Matplotlib and Seaborn (Bar chart, Line chart, Pie chart, Histogram, Boxplot, and Correlation Heatmap).


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sample Data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Marks': [85, 78, 92, 67, 88],
    'Subject': ['Math', 'Math', 'Math', 'Math', 'Math']
}
df = pd.DataFrame(data)
df
```

```python
# 1. Bar Chart (Name vs Marks)
plt.figure(figsize=(8, 5))
plt.bar(df['Name'], df['Marks'], color='skyblue', edgecolor='black')
plt.title('Student Marks')
plt.xlabel('Name')
plt.ylabel('Marks')
plt.show()
```

```python
# 2. Line Chart
plt.figure(figsize=(8, 5))
plt.plot(df['Name'], df['Marks'], marker='o', color='green', linewidth=2)
plt.title('Student Marks Trend')
plt.xlabel('Name')
plt.ylabel('Marks')
plt.grid(True)
plt.show()
```

```python
# 3. Pie Chart
plt.figure(figsize=(6, 6))
plt.pie(df['Marks'], labels=df['Name'], autopct='%1.1f%%', startangle=90)
plt.title('Marks Distribution')
plt.show()
```

```python
# 4. Histogram (Seaborn)
plt.figure(figsize=(8, 5))
sns.histplot(df['Marks'], bins=5, kde=True, color='purple')
plt.title('Distribution of Marks')
plt.show()
```

```python
# 5. Box Plot (to detect outliers)
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['Marks'], color='orange')
plt.title('Boxplot of Marks')
plt.show()
```

```python
# 6. Correlation Heatmap
df['Attendance'] = [90, 80, 95, 60, 85]
df['Assignment'] = [88, 70, 90, 55, 80]

plt.figure(figsize=(6, 5))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```
