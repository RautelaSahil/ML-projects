# Experiment 1: Create CSV Files & Load into Internal Data Structures

**Objective:** Demonstrate creating a CSV file programmatically and loading its contents into Python's native data structures (List of Lists and List of Dictionaries).


```python
import csv

# 1. Create and Save CSV File
data = [
    ["ID", "Name", "Marks"],
    [1, "Alice", 85],
    [2, "Bob", 78],
    [3, "Charlie", 92],
    [4, "David", 67]
]

with open("students.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file 'students.csv' created successfully!")
```

```python
# 2. Load as List of Lists
data_list = []
with open("students.csv", mode="r") as file:
    reader = csv.reader(file)
    for row in reader:
        data_list.append(row)

print("Loaded as List of Lists:")
for row in data_list:
    print(row)
```

```python
# 3. Load as List of Dictionaries
data_dict_list = []
with open("students.csv", mode="r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        data_dict_list.append(dict(row))

print("Loaded as List of Dictionaries:")
for item in data_dict_list:
    print(item)
```
