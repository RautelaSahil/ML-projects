# Experiment 3: Python Programming for Cleaning the Data

**Objective:** Perform fundamental data cleaning tasks on tabular data: removing duplicates, handling missing numerical & categorical values, whitespace stripping, and data type casting.


```python
import pandas as pd

# Create a sample dirty 'data.csv' to test with
sample_dirty = """Name,Age,Marks
 Alice ,20,85
Bob,,78
 Alice ,20,85
,22,
Charlie,21,92
"""
with open("data.csv", "w") as f:
    f.write(sample_dirty)

# Load dataset
df = pd.read_csv("data.csv")
print("Original Data:")
df
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Fill missing numerical values with mean
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Fill missing categorical values with mode
df["Name"] = df["Name"].fillna(df["Name"].mode()[0])

# Remove extra spaces from text columns
df["Name"] = df["Name"].str.strip()

# Convert data types if needed
df["Age"] = df["Age"].astype(int)

print("Cleaned Data:")
df
```

```python
# Save cleaned dataset
df.to_csv("cleaned_data.csv", index=False)
print("Data cleaning completed successfully and saved to 'cleaned_data.csv'!")
```
