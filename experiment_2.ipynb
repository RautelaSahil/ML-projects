# Experiment 2: Python Programming for Loading the Data

**Objective:** Explore different ways to load structured CSV data using standard `csv` library and `pandas`, and inspect rows and columns.


```python
import csv
import pandas as pd

# 1. Create and Save CSV File (if not already created)
data = [
    ["ID", "Name", "Marks"],
    [1, "Alice", 85],
    [2, "Bob", 78],
    [3, "Charlie", 92],
    [4, "David", 67]
]

with open("students.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file created.")
```

```python
# 2. Load CSV into List of Dictionaries
data_dict = []
with open("students.csv", mode="r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        data_dict.append(dict(row))

print("Data loaded from CSV (List of Dictionaries):")
for item in data_dict:
    print(item)
```

```python
# 3. Load CSV using pandas
df = pd.read_csv("students.csv")
print("Data loaded using pandas:")
df
```

```python
# 4. Access Individual Columns or Rows
print("Student Names:")
print(df["Name"])

print("
Second Row:")
print(df.iloc[1])
```
