# Experiment 1: Create CSV Files & Load into Internal Data Structures

# Experiment 5: Scikit-learn to Handle Regression

# Experiment 6: Non-Parametric Locally Weighted Regression (LWR)

**Objective:** Implement non-parametric Locally Weighted Regression (LWR) from scratch using a Gaussian kernel to fit non-linear data without making fixed parametric assumptions.


```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Sample Data (noisy sine wave)
np.random.seed(42)
X = np.linspace(1, 10, 100)
Y = np.sin(X) + np.random.randn(100) * 0.1

# Convert to matrix format for linear algebra
X = X[:, np.newaxis]                   # shape (100, 1)
X_b = np.hstack((np.ones_like(X), X))  # Add bias term -> shape (100, 2)
```

```python
# Step 2: Define Weight Function (Gaussian kernel)
def get_weights(X_query, X_train, tau):
    m = X_train.shape[0]
    weights = np.eye(m)
    for i in range(m):
        diff = X_train[i] - X_query
        weights[i, i] = np.exp(-(diff @ diff.T) / (2 * tau ** 2))
    return weights

# Step 3: Define LWR Prediction Function
def locally_weighted_regression(X_train, y_train, X_test, tau=0.5):
    m = X_test.shape[0]
    y_pred = np.zeros(m)
    
    for i in range(m):
        W = get_weights(X_test[i], X_train, tau)
        # theta = (X^T W X)^(-1) X^T W y
        XTWX = X_train.T @ W @ X_train
        XTWy = X_train.T @ W @ y_train
        theta = np.linalg.pinv(XTWX) @ XTWy
        y_pred[i] = X_test[i] @ theta
    return y_pred
```

```python
# Step 4: Fit and Visualize the Model
tau = 0.5  # Bandwidth parameter
y_pred = locally_weighted_regression(X_b, Y, X_b, tau)

plt.figure(figsize=(10, 6))
plt.scatter(X[:, 0], Y, label='Training Data', color='lightgray')
plt.plot(X[:, 0], y_pred, color='red', linewidth=2, label=f'LWR Prediction (tau={tau})')
plt.title(f'Locally Weighted Regression (tau = {tau})')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

# Experiment 7: Simple Linear Regression (Single Feature)

**Objective:** Perform Simple Linear Regression on the standard Diabetes dataset using a single predictive feature (`bmi`) to model disease progression.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1) Load data & select single feature ('bmi')
diabetes = load_diabetes()
X_all = diabetes.data
y = diabetes.target
feature_names = diabetes.feature_names

bmi_idx = feature_names.index('bmi')
X = X_all[:, [bmi_idx]]  # keep 2D shape: (n_samples, 1)

print("Feature chosen:", feature_names[bmi_idx])
print("X shape:", X.shape, " y shape:", y.shape)
```

```python
# 5) Predict & Evaluate
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Evaluation on Test Set")
print(f"MAE  : {mae:.3f}")
print(f"MSE  : {mse:.3f}")
print(f"RMSE : {rmse:.3f}")
print(f"R^2  : {r2:.3f}")
```

# Experiment 8: Multiple Linear Regression

**Objective:** Compare multiple linear regression models (2 features vs. all features) with feature scaling, multicollinearity assessment (VIF), residual diagnostics, and model serialization with `joblib`.


```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1) Load data & build DataFrame
diabetes = load_diabetes()
X_all = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
y = pd.Series(diabetes.target, name='target')

print("Dataset shape:", X_all.shape)
print("Feature names:", list(diabetes.feature_names))
X_all.head()
```

```python
# 2) Helper function for Model Fitting, Evaluation, and VIF
features_two = ['bmi', 'bp']         # Model A (two features)
features_all = list(X_all.columns)   # Model B (all features)

def fit_and_evaluate(feature_list, X_df, y, random_state=42):
    X = X_df[feature_list].values
    X_train, X_test, y_train, y_test = train_test_split(
        X, y.values, test_size=0.2, random_state=random_state
    )
    
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled  = scaler.transform(X_test)
    
    model = LinearRegression()
    model.fit(X_train_scaled, y_train)
    
    y_pred_train = model.predict(X_train_scaled)
    y_pred_test  = model.predict(X_test_scaled)
    
    def metrics(y_true, y_pred, p):
        mae = mean_absolute_error(y_true, y_pred)
        mse = mean_squared_error(y_true, y_pred)
        rmse = np.sqrt(mse)
        r2 = r2_score(y_true, y_pred)
        n = y_true.shape[0]
        adj_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1) if (n - p - 1) > 0 else np.nan
        return {'MAE': mae, 'MSE': mse, 'RMSE': rmse, 'R2': r2, 'Adj_R2': adj_r2}
    
    metrics_train = metrics(y_train, y_pred_train, p=X_train.shape[1])
    metrics_test  = metrics(y_test, y_pred_test, p=X_test.shape[1])
    
    coef_map = dict(zip(feature_list, model.coef_))
    residuals = y_test - y_pred_test
    
    def compute_vif(X_mat, feature_names):
        vif_values = []
        X_np = np.array(X_mat)
        p = X_np.shape[1]
        for i in range(p):
            X_i = X_np[:, i]
            X_other = np.delete(X_np, i, axis=1)
            if X_other.shape[1] == 0:
                r2 = 0.0
            else:
                lr = LinearRegression().fit(X_other, X_i)
                r2 = lr.score(X_other, X_i)
            vif = 1.0 / (1.0 - r2) if (1.0 - r2) > 1e-10 else np.inf
            vif_values.append(vif)
        return pd.DataFrame({'feature': feature_names, 'VIF': np.round(vif_values, 3)})
    
    vif_df = compute_vif(X_train_scaled, feature_list)
    
    return {
        'model': model, 'scaler': scaler, 'features': feature_list,
        'coef_map': coef_map, 'intercept': model.intercept_,
        'metrics_train': metrics_train, 'metrics_test': metrics_test,
        'X_train_scaled': X_train_scaled, 'X_test_scaled': X_test_scaled,
        'y_train': y_train, 'y_test': y_test, 'y_pred_test': y_pred_test,
        'residuals': residuals, 'vif': vif_df
    }
```

```python
# 3) Fit Model A (two features)
print("Fitting Model A (two features: bmi, bp)...")
res_a = fit_and_evaluate(features_two, X_all, y)
print("Intercept:", round(res_a['intercept'], 4))
print("Coefficients:", res_a['coef_map'])
print("Test metrics :", res_a['metrics_test'])
print("VIF:
", res_a['vif'])
```

```python
# 4) Fit Model B (all features)
print("Fitting Model B (all features)...")
res_b = fit_and_evaluate(features_all, X_all, y)
print("Intercept:", round(res_b['intercept'], 4))
print("Test metrics :", res_b['metrics_test'])
print("VIF:
", res_b['vif'])
```

```python
# 5) Compare Model A vs Model B
print("Comparison on TEST set:")
print(f"Model A (features={res_a['features']}): R2={res_a['metrics_test']['R2']:.4f}, AdjR2={res_a['metrics_test']['Adj_R2']:.4f}")
print(f"Model B (features=all):           R2={res_b['metrics_test']['R2']:.4f}, AdjR2={res_b['metrics_test']['Adj_R2']:.4f}")
```

```python
# 6) Diagnostic plots for Model B
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.scatter(res_b['y_test'], res_b['y_pred_test'], alpha=0.7)
minv = min(res_b['y_test'].min(), res_b['y_pred_test'].min())
maxv = max(res_b['y_test'].max(), res_b['y_pred_test'].max())
plt.plot([minv, maxv], [minv, maxv], linestyle='--', linewidth=1, color='red', label='ideal')
plt.xlabel('Actual target (test)')
plt.ylabel('Predicted target (test)')
plt.title('Actual vs Predicted (Model B)')
plt.legend()
plt.grid(alpha=0.3)

plt.subplot(1, 2, 2)
plt.scatter(res_b['y_pred_test'], res_b['residuals'], alpha=0.7)
plt.axhline(0, color='black', linestyle='--')
plt.xlabel('Predicted (test)')
plt.ylabel('Residuals (test)')
plt.title('Residuals vs Predicted (Model B)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 4))
plt.hist(res_b['residuals'], bins=20, edgecolor='k', alpha=0.7)
plt.title('Residuals distribution (Model B test set)')
plt.xlabel('Residual')
plt.ylabel('Frequency')
plt.grid(alpha=0.3)
plt.show()
```

```python
# 7) Save Model B & Example Prediction
joblib.dump({'model': res_b['model'], 'scaler': res_b['scaler'], 'features': res_b['features']},
            'linear_model_b_all_features.joblib')
print("Saved Model B to 'linear_model_b_all_features.joblib'")

example_raw = np.array([X_all.iloc[0].values])
scaled_example = res_b['scaler'].transform(example_raw)
pred_example = res_b['model'].predict(scaled_example)[0]
print("Example prediction (row 0) -> Actual:", float(y.iloc[0]), "| Predicted:", round(float(pred_example), 2))
```

# Experiment 9: Regression Model to Predict House Prices

**Objective:** Predict California housing prices using multiple regression models (Linear Regression, Ridge Regression, and Random Forest Regressor), optimize hyperparameters with GridSearchCV, and evaluate with 5-fold cross-validation.


```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import joblib
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# 1) Load dataset
data = fetch_california_housing()
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name='MedHouseVal')  # target units ~ 100k USD

print("Dataset shape:", X.shape)
print("Top correlations with target:
", X.join(y).corr()['MedHouseVal'].sort_values(ascending=False))
```

```python
# 2) Train/Test split & Scale
RANDOM_STATE = 42
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

def adjusted_r2(r2, n, p):
    return 1 - (1 - r2) * (n - 1) / (n - p - 1) if (n - p - 1) > 0 else np.nan

def evaluate_model(name, model, X_tr, X_te, y_tr, y_te):
    y_pred = model.predict(X_te)
    mae = mean_absolute_error(y_te, y_pred)
    mse = mean_squared_error(y_te, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_te, y_pred)
    adj = adjusted_r2(r2, y_te.shape[0], X_tr.shape[1])
    cv_scores = cross_val_score(model, X_tr, y_tr, scoring='neg_mean_squared_error', cv=5, n_jobs=-1)
    cv_rmse = np.sqrt(-cv_scores).mean()
    
    print(f"---- {name} ----")
    print(f"MAE: {mae:.4f} | RMSE: {rmse:.4f} | R2: {r2:.4f} | AdjR2: {adj:.4f} | CV RMSE: {cv_rmse:.4f}
")
    return {'name': name, 'mae': mae, 'mse': mse, 'rmse': rmse, 'r2': r2, 'adj_r2': adj, 'y_pred': y_pred}
```

```python
# 3) Baseline: Linear Regression
lin = LinearRegression().fit(X_train_scaled, y_train)
res_lin = evaluate_model("LinearRegression", lin, X_train_scaled, X_test_scaled, y_train, y_test)
```

```python
# 4) Ridge Regression (GridSearchCV)
gs_ridge = GridSearchCV(Ridge(random_state=RANDOM_STATE), {'alpha': [0.1, 1.0, 10.0, 50.0]},
                        scoring='neg_mean_squared_error', cv=5, n_jobs=-1)
gs_ridge.fit(X_train_scaled, y_train)
best_ridge = gs_ridge.best_estimator_
res_ridge = evaluate_model("Ridge (best)", best_ridge, X_train_scaled, X_test_scaled, y_train, y_test)
```

```python
# 5) Random Forest Regressor (GridSearchCV)
gs_rf = GridSearchCV(RandomForestRegressor(random_state=RANDOM_STATE),
                     {'n_estimators': [100, 200], 'max_depth': [None, 10, 20]},
                     scoring='neg_mean_squared_error', cv=3, n_jobs=-1)
gs_rf.fit(X_train, y_train)
best_rf = gs_rf.best_estimator_
res_rf = evaluate_model("RandomForest (best)", best_rf, X_train.values, X_test.values, y_train, y_test)
```

```python
# 6) Compare & Choose Best Model
results = pd.DataFrame([
    {'model': res_lin['name'], 'rmse': res_lin['rmse'], 'r2': res_lin['r2']},
    {'model': res_ridge['name'], 'rmse': res_ridge['rmse'], 'r2': res_ridge['r2']},
    {'model': res_rf['name'], 'rmse': res_rf['rmse'], 'r2': res_rf['r2']}
])
print("Summary comparison:
", results.sort_values('rmse'))

best_model_name = results.loc[results['rmse'].idxmin(), 'model']
print("
Best model on test RMSE:", best_model_name)

if best_model_name.startswith("RandomForest"):
    best_model, y_pred = best_rf, res_rf['y_pred']
    joblib.dump({'model': best_model, 'scaler': None, 'features': list(X.columns)}, "best_house_price_model.joblib")
elif best_model_name.startswith("Ridge"):
    best_model, y_pred = best_ridge, res_ridge['y_pred']
    joblib.dump({'model': best_model, 'scaler': scaler, 'features': list(X.columns)}, "best_house_price_model.joblib")
else:
    best_model, y_pred = lin, res_lin['y_pred']
    joblib.dump({'model': best_model, 'scaler': scaler, 'features': list(X.columns)}, "best_house_price_model.joblib")
print("Saved best model to 'best_house_price_model.joblib'")
```

# Experiment 10: FIND-S Algorithm

**Objective:** Implement the FIND-S concept learning algorithm from machine learning to find the maximally specific hypothesis consistent with positive training examples.


```python
import pandas as pd
import os

# 1. Generate sample play_tennis.csv if not exists
sample = """Outlook,Temperature,Humidity,Wind,PlayTennis
Sunny,Hot,High,Weak,No
Sunny,Hot,High,Strong,No
Overcast,Hot,High,Weak,Yes
Rain,Mild,High,Weak,Yes
Rain,Cool,Normal,Weak,Yes
Rain,Cool,Normal,Strong,No
Overcast,Cool,Normal,Strong,Yes
Sunny,Mild,High,Weak,No
Sunny,Cool,Normal,Weak,Yes
Rain,Mild,Normal,Weak,Yes
Sunny,Mild,Normal,Strong,Yes
Overcast,Mild,High,Strong,Yes
Overcast,Hot,Normal,Weak,Yes
Rain,Mild,High,Strong,No
"""
with open('play_tennis.csv', 'w', newline='') as f:
    f.write(sample)

df = pd.read_csv('play_tennis.csv')
print("Dataset loaded:")
df
```

```python
# 2. Define FIND-S Function
def find_s(df, target_col, positive_label=None, verbose=True):
    df = df.copy()
    if positive_label is None:
        unique = df[target_col].unique().tolist()
        candidates = [u for u in unique if str(u).strip().lower() in ('yes', 'y', 'positive', '1', 'true', 't')]
        positive_label = candidates[0] if candidates else unique[0]
        
    features = [c for c in df.columns if c != target_col]
    h = [None] * len(features)  # None means Ø (most specific)
    steps = [("Initial", h.copy())]
    
    if verbose:
        print("Initial hypothesis:", ["Ø" for _ in features])
        
    for idx, row in df.iterrows():
        row_label = str(row[target_col]).strip().lower()
        if row_label == str(positive_label).strip().lower():
            for i, feat in enumerate(features):
                value = row[feat]
                if h[i] is None:
                    h[i] = value
                elif h[i] != value:
                    h[i] = '?'
            steps.append((f"After positive example row {idx+1}", h.copy()))
            if verbose:
                readable = [(h_i if h_i is not None else 'Ø') for h_i in h]
                print(f"After positive example {idx+1}: {[row[f] for f in features]} -> hypothesis: {readable}")
                
    return features, h, positive_label, steps
```

```python
# 3. Define Prediction and Evaluation Functions
def predict_with_h(h, features, instance):
    for i, feat in enumerate(features):
        constraint = h[i]
        val = instance[feat]
        if constraint == '?':
            continue
        if constraint is None or constraint != val:
            return False
    return True

def evaluate_hypothesis(h, features, df, target_col, positive_label=None):
    if positive_label is None:
        unique = df[target_col].unique().tolist()
        candidates = [u for u in unique if str(u).strip().lower() in ('yes', 'y', 'positive', '1', 'true', 't')]
        positive_label = candidates[0] if candidates else unique[0]
        
    preds = []
    TP = TN = FP = FN = 0
    for _, row in df.iterrows():
        pred_pos = predict_with_h(h, features, row)
        actual_pos = (str(row[target_col]).strip().lower() == str(positive_label).strip().lower())
        preds.append('Yes' if pred_pos else 'No')
        if pred_pos and actual_pos:
            TP += 1
        elif pred_pos and not actual_pos:
            FP += 1
        elif not pred_pos and actual_pos:
            FN += 1
        else:
            TN += 1
            
    total = len(df)
    accuracy = (TP + TN) / total if total > 0 else 0.0
    return preds, accuracy, TP, TN, FP, FN
```

```python
# 4. Run FIND-S on dataset
features, h, pos_label, _ = find_s(df, target_col="PlayTennis", verbose=True)

print("
Final hypothesis (feature -> constraint):")
readable = [(h_i if h_i is not None else 'Ø') for h_i in h]
for feat, constraint in zip(features, readable):
    print(f"  {feat} -> {constraint}")
    
preds, acc, TP, TN, FP, FN = evaluate_hypothesis(h, features, df, target_col="PlayTennis")
print("
Evaluation on dataset:")
print(f"  Positive label used: {pos_label}")
print(f"  Accuracy: {acc:.3f} (TP={TP}, TN={TN}, FP={FP}, FN={FN})")
```
